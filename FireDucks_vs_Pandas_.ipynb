```python
import pandas as pd
import numpy as np

# Create 10 Million rows
df = pd.DataFrame({
    'customer_id': np.random.randint(1, 1_000_000, 10_000_000),
    'transaction_amount': np.random.rand(10_000_000) * 5000,  # transactions up to $5000
    'product_category': np.random.choice(['electronics', 'home_goods', 'clothing', 'sports'], 10_000_000),
    'purchase_date': pd.date_range(start='2022-01-01', periods=10_000_000, freq='s')
})

# Update the purchase_date to more relevant dates for visualization
df['purchase_date'] = pd.to_datetime('2023-01-01') + pd.to_timedelta(np.random.randint(0, 365*24*3600, size=10_000_000), unit='s')

# Save to CSV
df.to_csv('large_eda_dataset.csv', index=False)

print("EDA Dataset with 10 Million Rows Generated Successfully!")
```

Filtering Rows

```python
import pandas as pd
import fireducks.pandas as fd
import time

# Load dataset
df_pandas = pd.read_csv('large_eda_dataset.csv')
df_fireducks = fd.read_csv('large_eda_dataset.csv')

# Filtering rows by condition
def filter_rows_pandas(df):
    return df[df['transaction_amount'] > 2000]

def filter_rows_fireducks(df):
    return df[df['transaction_amount'] > 2000]

# Benchmarking
start_time = time.time()
filtered_df_pandas = filter_rows_pandas(df_pandas)
end_time = time.time()
print(f"Pandas Filtering Time: {end_time - start_time:.4f} seconds")

start_time = time.time()
filtered_df_fireducks = filter_rows_fireducks(df_fireducks)
end_time = time.time()
print(f"FireDucks Filtering Time: {end_time - start_time:.4f} seconds")
```

GroupBy & Aggregation

```python
# Load dataset
df_pandas = pd.read_csv('large_eda_dataset.csv')
df_fireducks = fd.read_csv('large_eda_dataset.csv')

# GroupBy and aggregation
def groupby_pandas(df):
    return df.groupby('product_category')['transaction_amount'].sum()

def groupby_fireducks(df):
    return df.groupby('product_category')['transaction_amount'].sum()

# Benchmarking
start_time = time.time()
result_pandas = groupby_pandas(df_pandas)
end_time = time.time()
print(f"Pandas GroupBy Time: {end_time - start_time:.4f} seconds")

start_time = time.time()
result_fireducks = groupby_fireducks(df_fireducks)
end_time = time.time()
print(f"FireDucks GroupBy Time: {end_time - start_time:.4f} seconds")
```

Data Merging

```python
import pandas as pd
import fireducks.pandas as fd
import time

# Load dataset
df1_pandas = pd.read_csv('large_eda_dataset.csv')
df2_pandas = df1_pandas.sample(frac=0.5)  # Create another sample dataset for merging

df1_fireducks = fd.read_csv('large_eda_dataset.csv')
df2_fireducks = df1_fireducks.sample(frac=0.5)

# Merging datasets
def merge_pandas(df1, df2):
    return pd.merge(df1, df2, on='customer_id')

def merge_fireducks(df1, df2):
    return fd.merge(df1, df2, on='customer_id')

# Benchmarking
start_time = time.time()
merged_pandas = merge_pandas(df1_pandas, df2_pandas)
end_time = time.time()
print(f"Pandas Merge Time: {end_time - start_time:.4f} seconds")

start_time = time.time()
merged_fireducks = merge_fireducks(df1_fireducks, df2_fireducks)
end_time = time.time()
print(f"FireDucks Merge Time: {end_time - start_time:.4f} seconds")
```

Visualization Preparation

```python
import pandas as pd
import fireducks.pandas as fd
import time

# Load dataset
df_pandas = pd.read_csv('large_eda_dataset.csv')
df_fireducks = fd.read_csv('large_eda_dataset.csv')

# Summarizing statistics for visualization
def prepare_pandas(df):
    return df.groupby('purchase_date')['transaction_amount'].mean().reset_index()

def prepare_fireducks(df):
    return df.groupby('purchase_date')['transaction_amount'].mean().reset_index()

# Benchmarking
start_time = time.time()
result_pandas = prepare_pandas(df_pandas)
end_time = time.time()
print(f"Pandas Preparation Time: {end_time - start_time:.4f} seconds")

start_time = time.time()
result_fireducks = prepare_fireducks(df_fireducks)
end_time = time.time()
print(f"FireDucks Preparation Time: {end_time - start_time:.4f} seconds")
```
